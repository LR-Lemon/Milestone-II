```python
!pip install -q yfinance pyarrow

import pandas as pd
import yfinance as yf


# ==========================================
# 1. S&P 500
# ==========================================

sp500 = yf.download(
    "^GSPC",
    start="2000-01-01",
    end="2026-01-01",
    interval="1d",
    auto_adjust=False,
    progress=False
)

# Flatten columns if needed
if isinstance(sp500.columns, pd.MultiIndex):
    sp500.columns = sp500.columns.get_level_values(0)

sp500 = sp500.reset_index()

sp500.to_parquet(
    "sp500_2000_2025.parquet",
    engine="pyarrow",
    index=False
)


# ==========================================
# 2. CBOE VIX
# ==========================================

vix_url = "https://cdn.cboe.com/api/global/us_indices/daily_prices/VIX_History.csv"

vix = pd.read_csv(vix_url)

vix["DATE"] = pd.to_datetime(vix["DATE"])

vix = vix[
    (vix["DATE"] >= "2000-01-01") &
    (vix["DATE"] <= "2025-12-31")
].copy()

vix.to_parquet(
    "vix_2000_2025.parquet",
    engine="pyarrow",
    index=False
)


print("S&P 500 shape:", sp500.shape)
print("VIX shape:", vix.shape)

display(sp500.head())
display(vix.head())
```

```python
print("S&P 500 SHAPE")
print(sp500.shape)

print("\nS&P 500 COLUMNS")
print(sp500.columns.tolist())

print("\nS&P 500 DTYPES")
print(sp500.dtypes)

print("\nS&P 500 DATE RANGE")
print(sp500["Date"].min(), "to", sp500["Date"].max())

print("\nS&P 500 MISSING VALUES")
print(sp500.isna().sum())

print("\nS&P 500 DUPLICATE ROWS")
print(sp500.duplicated().sum())

print("\nS&P 500 DUPLICATE DATES")
print(sp500["Date"].duplicated().sum())

print("\nS&P 500 DESCRIPTIVE STATISTICS")
display(sp500.describe().T)
```

```python
print("VIX SHAPE")
print(vix.shape)

print("\nVIX COLUMNS")
print(vix.columns.tolist())

print("\nVIX DTYPES")
print(vix.dtypes)

print("\nVIX DATE RANGE")
print(vix["DATE"].min(), "to", vix["DATE"].max())

print("\nVIX MISSING VALUES")
print(vix.isna().sum())

print("\nVIX DUPLICATE ROWS")
print(vix.duplicated().sum())

print("\nVIX DUPLICATE DATES")
print(vix["DATE"].duplicated().sum())

print("\nVIX DESCRIPTIVE STATISTICS")
display(vix.describe().T)
```

```python
sp500_dates = set(sp500["Date"])
vix_dates = set(vix["DATE"])

only_sp500 = sorted(sp500_dates - vix_dates)
only_vix = sorted(vix_dates - sp500_dates)

print("Dates only in S&P 500:", len(only_sp500))
print(only_sp500)

print("\nDates only in VIX:", len(only_vix))
print(only_vix)
```

```python
vix_extra = vix[vix["DATE"].isin(only_vix)].copy()

display(vix_extra)
```

```python
vix_extra.describe()
```

```python
market = sp500.merge(
    vix,
    left_on="Date",
    right_on="DATE",
    how="inner",
    suffixes=("_SP500", "_VIX")
)

market.shape
```

```python
print("Rows:", len(market))
print("Start:", market["Date"].min())
print("End:", market["Date"].max())
print("Duplicate dates:", market["Date"].duplicated().sum())
print("Missing values:", market.isna().sum().sum())
```

```python
market.head()
```

```python
market.tail()
```

```python
market = market.rename(columns={
    "Adj Close": "SP500_Adj_Close",
    "Close": "SP500_Close",
    "High": "SP500_High",
    "Low": "SP500_Low",
    "Open": "SP500_Open",
    "Volume": "SP500_Volume",

    "OPEN": "VIX_Open",
    "HIGH": "VIX_High",
    "LOW": "VIX_Low",
    "CLOSE": "VIX_Close"
})
```

```python
market = market.drop(columns="DATE")
```

```python
market.columns.tolist()
```

```python
market = market[
    [
        "Date",
        "SP500_Open",
        "SP500_High",
        "SP500_Low",
        "SP500_Close",
        "SP500_Adj_Close",
        "SP500_Volume",
        "VIX_Open",
        "VIX_High",
        "VIX_Low",
        "VIX_Close"
    ]
]
```

```python
market.head()
```

```python
market.to_parquet(
    "market_2000_2025.parquet",
    engine="pyarrow",
    index=False
)
```
